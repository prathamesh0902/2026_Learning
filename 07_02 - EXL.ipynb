```python
# Customer table
# customer_id  |  email  |  phone |  updated_at |
# email is updated per customer_id
# find latest record per customer

select *,
    ROW_NUMBER() OVER(partition by customer_id order by updated_at desc) as latest_email_count
    from customers
where latest_email_count = 1
```
