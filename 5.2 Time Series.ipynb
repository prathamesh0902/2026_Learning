## Time Series

ARIMA, SARIMA, and SARIMAX are all time series forecasting models used to predict future values based on past data. Each model builds on the previous one, adding complexity to handle specific types of data patterns. Here's a breakdown of the differences:

### 1. **ARIMA (AutoRegressive Integrated Moving Average)**
ARIMA is a basic time series forecasting model that combines three components:

- **AR (AutoRegressive)**: This component models the relationship between an observation and a number of lagged observations (i.e., previous values in the series). It's essentially using past values to predict future ones.
  
- **I (Integrated)**: This component involves differencing the data to make it stationary, meaning the mean, variance, and autocorrelation structure do not change over time. The integration step removes trends and seasonality.
  
- **MA (Moving Average)**: This component models the relationship between an observation and a residual error from a moving average model applied to lagged observations. It captures the noise in the time series.

**ARIMA Model Notation**:
- ARIMA(p, d, q)
  - **p**: Number of lag observations included in the model (AR order).
  - **d**: Number of times the raw observations are differenced (degree of differencing).
  - **q**: Size of the moving average window (MA order).

### 2. **SARIMA (Seasonal AutoRegressive Integrated Moving Average)**
SARIMA extends ARIMA to handle **seasonal** patterns in the data. It includes seasonal components in addition to the regular ARIMA components.

SARIMA includes the following components:

- **Seasonal AR (AutoRegressive)**: Similar to AR, but applied to seasonal lagged observations (e.g., data points from the same season in previous cycles).
  
- **Seasonal I (Integrated)**: Seasonal differencing to remove seasonality (e.g., subtracting the value from the same season last year).
  
- **Seasonal MA (Moving Average)**: Similar to MA, but applied to seasonal errors.

**SARIMA Model Notation**:
- SARIMA(p, d, q)(P, D, Q)[m]
  - **(p, d, q)**: Non-seasonal parameters (as in ARIMA).
  - **(P, D, Q)**: Seasonal parameters.
  - **m**: The number of observations per season (e.g., `m=12` for monthly data with a yearly seasonal pattern).

### 3. **SARIMAX (Seasonal AutoRegressive Integrated Moving Average with Exogenous Regressors)**
SARIMAX is an extension of SARIMA that allows for the inclusion of **exogenous variables** (external factors) that can influence the forecast. These variables are not part of the time series itself but can help improve the model's accuracy.

SARIMAX includes all the components of SARIMA, plus:

- **Exogenous Regressors (X)**: Additional external variables (e.g., economic indicators, weather data) that are believed to have an influence on the target variable.

**SARIMAX Model Notation**:
- SARIMAX(p, d, q)(P, D, Q)[m] + X
  - **X**: Represents the exogenous variables added to the model.

### Summary of Differences:

- **ARIMA**: Used for non-seasonal data and does not include any external variables.
- **SARIMA**: Adds seasonal components to ARIMA, making it suitable for data with seasonal patterns.
- **SARIMAX**: Builds on SARIMA by allowing the inclusion of exogenous variables, making it suitable for complex time series data that are influenced by external factors.

### When to Use Each Model:

- **ARIMA**: Use when your time series data does not exhibit seasonality or is stationary after differencing.
- **SARIMA**: Use when your data shows seasonal patterns (e.g., monthly sales data that peaks in December).
- **SARIMAX**: Use when your data is influenced by external factors (e.g., sales data influenced by marketing spend or economic indicators).

Each of these models is powerful in its own context, and the choice between them depends on the specific characteristics of the time series data you are analyzing.

## 1. AR (AutoRegressive) Model — 5 Key Points

An **AutoRegressive (AR)** model predicts the current value using its own past values.

### Example

$$[
Y_t = c + \phi_1Y_{t-1} + \phi_2Y_{t-2} + \cdots + \phi_pY_{t-p} + \epsilon_t
]$$

where:

* (Y_t) = current value
* (Y_{t-1},Y_{t-2}) = previous observations
* (p) = number of lags
* (\epsilon_t) = random error

### Key Points

1. Current value depends on previous values of the same series.
2. Parameter **p** denotes how many past observations are used.
3. Works best when the series is **stationary**.
4. Captures persistence and momentum in data.
5. PACF helps determine the optimal value of **p**.

### Example

Stock price today may depend on stock prices from the last 2 days:

$$[
Y_t = 0.7Y_{t-1} + 0.2Y_{t-2} + \epsilon_t
]$$

This is an AR(2) model.

---

## 2. MA (Moving Average) Model — 5 Key Points

An **MA model** predicts current values using past forecast errors.

### Example

$$[
Y_t = \mu + \epsilon_t + \theta_1\epsilon_{t-1} + \theta_2\epsilon_{t-2}
]$$

where:

* (\epsilon_t) = current random shock
* (\epsilon_{t-1}) = previous shock
* (q) = number of lagged errors

### Key Points

1. Current value depends on past errors, not past observations.
2. Parameter **q** represents number of previous error terms.
3. Useful when sudden shocks affect future values.
4. Captures short-term randomness.
5. ACF helps determine the value of **q**.

### Example

Suppose a sales spike occurred due to a festival.

Today's sales may still reflect part of yesterday's unexpected spike:

$$[
Y_t = \mu + \epsilon_t + 0.5\epsilon_{t-1}
]$$

This is an MA(1) model.

---

## 3. I (Integrated) Model — 5 Key Points

The **Integrated (I)** part refers to differencing the series to make it stationary.

### First Difference

$$[
Y'*t = Y_t - Y*{t-1}
]$$

### Key Points

1. Removes trend from data.
2. Converts non-stationary series into stationary series.
3. Parameter **d** indicates number of differencing operations.
4. Necessary because AR and MA require stationarity.
5. Determined using plots and stationarity tests.


# 4. ARIMA Model

ARIMA combines:

* AR → Past values
* I → Differencing
* MA → Past errors

[
ARIMA(p,d,q)
]

where:

* p = AR order
* d = differencing order
* q = MA order

Example:

[
ARIMA(2,1,1)
]

means:

* Use 2 lagged observations
* Difference once
* Use 1 lagged error

---

# 5. What Does ACF Mean?

## ACF = AutoCorrelation Function

Measures correlation between current observation and its lagged values.

[
Corr(Y_t,Y_{t-k})
]

where (k) is the lag.

### Interpretation

| Lag | ACF  |
| --- | ---- |
| 1   | 0.85 |
| 2   | 0.75 |
| 3   | 0.65 |

High values indicate strong dependence on previous observations.

### Uses

1. Detect seasonality.
2. Detect trend.
3. Identify MA(q).
4. Understand persistence.
5. Check residual independence.

---

# 6. What Does PACF Mean?

## PACF = Partial AutoCorrelation Function

Measures the correlation between:

$$[
Y_t \text{ and } Y_{t-k}
]$$

after removing the effects of intermediate lags.

### Example

Suppose:

$$[
Y_t \leftrightarrow Y_{t-1} \leftrightarrow Y_{t-2}
]$$

The correlation between (Y_t) and (Y_{t-2}) may exist only because both are related through (Y_{t-1}).

PACF removes that indirect effect.

### Uses

1. Identify AR(p).
2. Determine significant lags.
3. Remove indirect correlations.
4. Model selection.
5. Understand direct dependency.

---

# 7. How ACF and PACF Select ARIMA Parameters

| Model     | ACF Pattern           | PACF Pattern          |
| --------- | --------------------- | --------------------- |
| AR(p)     | Tails off gradually   | Cuts off after p lags |
| MA(q)     | Cuts off after q lags | Tails off gradually   |
| ARMA(p,q) | Tails off             | Tails off             |

### One-Line Summary

**AR uses past values, MA uses past errors, I removes trend through differencing, PACF helps find p, ACF helps find q, and time-series analysis consists of making the data stationary, identifying lag relationships, fitting a model, validating residuals, and forecasting future values.**


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from statsmodels.tsa.seasonal import seasonal_decompose
from statsmodels.graphics.tsaplots import plot_acf, plot_pacf
from statsmodels.tsa.stattools import adfuller
from statsmodels.tsa.arima.model import ARIMA

# Create synthetic time series data
np.random.seed(42)
num_periods = 36  # Number of time periods
t = pd.date_range(start='2022-01-01', periods=num_periods, freq='M')  # Create a DatetimeIndex
noise = np.random.normal(0, 1, num_periods)
sales = 100 + 2 * np.arange(1, num_periods + 1) + 5 * np.sin(2 * np.pi * np.arange(1, num_periods + 1) / 12) + noise

# Create a DataFrame
df = pd.DataFrame({'timestamp': pd.date_range(start='2022-01-01', periods=num_periods, freq='ME'),
                   'sales': sales}, index=t)

# Visualize the synthetic time series data
plt.figure(figsize=(10, 6))
plt.plot(df['timestamp'], df['sales'])
plt.xlabel('Timestamp')
plt.ylabel('Sales')
plt.title('Synthetic Monthly Sales Time Series')
plt.show()



# Decompose the time series
decomposition = seasonal_decompose(df['sales'], model='additive')
trend = decomposition.trend
seasonal = decomposition.seasonal
residual = decomposition.resid

# Plot the decomposed components
plt.figure(figsize=(10, 8))
plt.subplot(411)
plt.plot(df['timestamp'], df['sales'], label='Original')
plt.legend()

plt.subplot(412)
plt.plot(df['timestamp'], trend, label='Trend')
plt.legend()

plt.subplot(413)
plt.plot(df['timestamp'], seasonal, label='Seasonality')
plt.legend()

plt.subplot(414)
plt.plot(df['timestamp'], residual, label='Residuals')
plt.legend()

plt.tight_layout()
plt.show()

# Autocorrelation and Partial Autocorrelation
plot_acf(df['sales'])
plot_pacf(df['sales'])
plt.show()

# Stationarity Check (Augmented Dickey-Fuller Test)
result = adfuller(df['sales'])
print('ADF Statistic:', result[0])
print('p-value:', result[1])

# Modeling - ARIMA
model = ARIMA(df['sales'], order=(1, 1, 1))  # Example order (p, d, q)
results = model.fit()

# Forecasting
forecast_steps = 6  # Example number of forecast steps
forecast = results.forecast(steps=forecast_steps)

# Visualize Results
plt.plot(df['timestamp'], df['sales'], label='Actual')
plt.plot(pd.date_range(start=df['timestamp'].iloc[-1], periods=forecast_steps+1, freq='MS')[1:], forecast, color='red', label='Forecast')
plt.xlabel('Timestamp')
plt.ylabel('Sales')
plt.title('Sales Forecast')
plt.legend()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

from statsmodels.tsa.arima_process import ArmaProcess
from statsmodels.tsa.arima.model import ARIMA
from statsmodels.graphics.tsaplots import plot_acf, plot_pacf

np.random.seed(42)
```

```python

##############################################################
# Step 1 : Generate ARMA(3,4) stationary process
##############################################################

# AR coefficients
phi = np.array([0.6, -0.45, 0.20])

# MA coefficients
theta = np.array([0.5, 0.3, -0.2, 0.1])

# statsmodels convention:
# AR = [1,-phi1,-phi2,...]
ar = np.r_[1, -phi]

# MA = [1,theta1,theta2,...]
ma = np.r_[1, theta]

arma = ArmaProcess(ar, ma)

stationary_series = arma.generate_sample(300)

##############################################################
# Step 2 : Integrate twice (d=2)
##############################################################

# First integration
integrated_once = np.cumsum(stationary_series)

# Second integration
integrated_twice = np.cumsum(integrated_once)

##############################################################
# Plot Original ARIMA(3,2,4)
##############################################################

plt.figure(figsize=(12,4))
plt.plot(integrated_twice)
plt.title("Synthetic ARIMA(3,2,4) Series")
plt.grid(True)
plt.show()

##############################################################
# Step 3 : Difference Once
##############################################################

diff1 = np.diff(integrated_twice)

plt.figure(figsize=(12,4))
plt.plot(diff1)
plt.title("First Difference")
plt.grid(True)
plt.show()

##############################################################
# Step 4 : Difference Twice
##############################################################

diff2 = np.diff(diff1)

plt.figure(figsize=(12,4))
plt.plot(diff2)
plt.title("Second Difference (Stationary)")
plt.grid(True)
plt.show()

##############################################################
# Step 5 : ACF & PACF
##############################################################

fig, ax = plt.subplots(1,2,figsize=(14,4))

plot_acf(diff2, lags=10, ax=ax[0])
plot_pacf(diff2, lags=10, ax=ax[1])

plt.show()

##############################################################
# Step 6 : Fit ARIMA(3,2,4)
##############################################################

model = ARIMA(integrated_twice, order=(3,2,4))
results = model.fit()

print(results.summary())

##############################################################
# Step 7 : Fitted Values
##############################################################

plt.figure(figsize=(12,4))

plt.plot(integrated_twice,label="Original")
plt.plot(results.fittedvalues,
         color="red",
         label="Fitted")

plt.legend()
plt.title("Original vs Fitted")
plt.grid(True)
plt.show()

##############################################################
# Step 8 : Residual Plot
##############################################################

residuals = results.resid

plt.figure(figsize=(12,4))
plt.plot(residuals)
plt.title("Residuals")
plt.grid(True)
plt.show()

##############################################################
# Residual ACF
##############################################################

plot_acf(residuals, lags=30)
plt.show()

##############################################################
# Step 9 : Forecast
##############################################################

forecast = results.forecast(20)

plt.figure(figsize=(12,4))

plt.plot(np.arange(len(integrated_twice)),
         integrated_twice,
         label="Observed")

plt.plot(np.arange(len(integrated_twice),
                   len(integrated_twice)+20),
         forecast,
         color="red",
         label="Forecast")

plt.legend()
plt.title("Forecast")
plt.grid(True)
plt.show()
```
