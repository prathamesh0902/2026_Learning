**📊 Data Scaling Techniques – Comparison Table**

| Technique                               | Formula                                      | Output Range      | Best Use Case                               | Handles Outliers | Key Advantage                    | Limitation                   |                                        |                               |                              |                                              |
| --------------------------------------- | -------------------------------------------- | ----------------- | ------------------------------------------- | ---------------- | -------------------------------- | ---------------------------- | -------------------------------------- | ----------------------------- | ---------------------------- | -------------------------------------------- |
| **Min-Max Scaling**                     | (X' = \frac{X - X_{min}}{X_{max} - X_{min}}) | [0, 1]            | When bounds are known, KNN, Neural Networks | ❌ No             | Preserves relative relationships | Highly sensitive to outliers |                                        |                               |                              |                                              |
| **Standardization (Z-score)**           | (X' = \frac{X - \mu}{\sigma})                | Mean = 0, Std = 1 | Linear models, SVM, PCA                     | ⚠️ Moderate      | Centers data, widely used        | Affected by extreme outliers |                                        |                               |                              |                                              |
| **Robust Scaling**                      | (X' = \frac{X - Median}{IQR})                | No fixed range    | Data with outliers                          | ✅ Yes            | Resistant to outliers            | Doesn’t bound values         |                                        |                               |                              |                                              |
| **Max Absolute Scaling**                | (X' = \frac{X}{                 X_{max}           })                                          | [-1, 1]          | Sparse data, text features       | ⚠️ Moderate                  | Preserves sparsity (zeros remain zero) | Sensitive to large max values |                              |                                              |
| **Log Transformation**                  | (X' = \log(X + 1))                           | Depends on data   | Right-skewed data                           | ✅ Yes            | Reduces skewness                 | Only works for positive data |                                        |                               |                              |                                              |
| **Power Transformation**                | Box-Cox / Yeo-Johnson                        | Approx normal     | Making data Gaussian                        | ✅ Yes            | Stabilizes variance              | More complex                 |                                        |                               |                              |                                              |
| **Unit Vector Scaling (Normalization)** | (X' = \frac{X}{                                                X                                                              })                               | Unit norm (length = 1)       | Text data, cosine similarity           | N/A                           | Useful for similarity models | Not useful for feature scaling in regression |

### Chi Squared

```python
import numpy as np

vounts = np.array([25, 250, 90, 15])

exp_per = [0.05, 0.38, 0.02, 0.55]
exp_freq = 180
exp_arr = exp_per * np.array(exp_freq)
```

```python
import numpy as np
from scipy.stats import chisquare
chi2_stat, p_value = chisquare(f_obs=vounts, f_exp=exp_arr)
print(f"Chi-squared Statistic: {chi2_stat:.4f}")
print(f"P-value: {p_value:.4f}")
```

```python
observed = np.array([25, 30, 20, 25])   # Observed frequencies
expected = np.array([20, 25, 25, 30])   # Expected frequencies

# Chi-squared test
chi2_stat, p_value = chisquare(f_obs=observed, f_exp=expected)

print(f"Chi-squared Statistic: {chi2_stat:.4f}")
print(f"P-value: {p_value:.4f}")
```

```python
import numpy as np
from scipy.stats import chisquare

# Observed counts
counts = np.array([25, 250, 90, 15])

# Expected percentages (in decimal form)
exp_per = np.array([0.05, 0.38, 0.02, 0.55])

# Match expected total to observed total
total_obs = counts.sum()
exp_arr = exp_per * total_obs  # element-wise multiplication

# Chi-squared test
chi2_stat, p_value = chisquare(f_obs=counts, f_exp=exp_arr)

print(f"Chi-squared Statistic: {chi2_stat:.4f}")
print(f"P-value: {p_value:.4f}")
```
