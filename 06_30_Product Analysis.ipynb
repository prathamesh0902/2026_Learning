````python
import pandas as pd
import matplotlib.pyplot as plt
from scipy import stats
import numpy as np
np.random.seed(42)

n1, n2 = 1000, 1000                     # sample sizes
X1, X2 = 120, 100                     # number of successes
p1, p2 = X1/n1, X2/n2               # sample proportion

π1, π2 = 0, 0                       # hypothesized population proportion
````

````python
confidence = 0.95
alpha = 1 - confidence
prob = 1 - alpha / 2

p_bar = (X1 + X2) / (n1 + n2)

# Standard Error
SE = np.sqrt(p_bar * (1 - p_bar) * (1 / n1 + 1 / n2))

# Z-score
z_score = stats.norm.ppf(prob)
z_value = ((p1 - p2) - (π1 - π2))/ SE
integral = stats.norm.cdf(abs(z_value))

p_value =  2 * (1 - integral)
print(f"p-value: {p_value* 100:.2f} %")
if p_value < alpha:
    print("Reject the Null Hypothesis")
else:
    print("Accept the Null Hypothesis")
````
