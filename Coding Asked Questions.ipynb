```python
# Given list
numbers = [1, 2, 2, 3, 3, 3, 5, 5]

# Empty dictionary to store frequencies
freq = {}

# Count each number
for num in numbers:
    freq[num] = freq.get(num, 0) + 1  # get() avoids KeyError

print(freq)  # {1: 1, 2: 2, 3: 3, 5: 2}
```

```python
numbers = [1, 2, 2, 3, 3, 3, 5, 5]
freq = {}

def count_freq(list):
    for num in list:
        if num in freq:
            freq[num] += 1
        else:
            freq[num] = 1

count_freq(numbers)
print(freq)
```

```python
5 in freq, freq[5]
```

```python
result = dict.fromkeys(numbers)
print(f"Result before: {result}")
for num in numbers:
    if result[num] == None:
        result[num] = 1
    else:
        result[num] += 1
print(f"Result after: {result}")
```

```python
## Q1. Which of the following is correct?

A: df[df['A'] > 2 and df['B']<5]
B: df[(df['A'] > 2) & (df['B']<5)]
C: df[df['A'] > 2 & df['B']<5]
D: df.filter[df['A'] > 2, df['B']<5]

**Answer:** ✅ **B. Why parentheses are needed**
& has higher precedence than comparison operators, so without parentheses,
df['A'] > 2 & df['B'] < 5 will be interpreted incorrectly and may throw an error.
```

```python
## Q5. What is the output?

import pandas as pd

df = pd.DataFrame({
    'grp': ['X', 'X', 'Y'],
    'val': [10, 20, 30]
})

print(df.groupby('grp')['val'].apply(lambda x: x-x.mean()))
```
